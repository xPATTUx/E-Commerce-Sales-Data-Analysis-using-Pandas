```python
# ============================================================
# 1. IMPORT LIBRARIES & LOAD DATASET
# ============================================================

import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv("Day8_Ecommerce_Sales_Dataset.csv")

# Display dataset
df.head()
```

```python
# ============================================================
# 2. DATASET EXPLORATION & BASIC STATISTICS
# ============================================================

print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nIndex:")
print(df.index)

print("\nData Types:")
print(df.dtypes)

print("\nDataset Information:")
df.info()

print("\nFirst 5 Rows:")
display(df.head())

print("\nLast 5 Rows:")
display(df.tail())

print("\nRandom Sample:")
display(df.sample(5))

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nDescriptive Statistics:")
display(df.describe())
```

```python
# ============================================================
# 3. DATA SELECTION, FILTERING & SORTING
# ============================================================

# Select specific columns
display(df[["Order_ID", "City", "Category", "Product", "Total_Sales"]].head())

# Select rows using iloc
display(df.iloc[0:10])

# Orders with sales greater than ₹5,000
high_value_orders = df[df["Total_Sales"] > 5000]
print("Orders with Sales > ₹5,000:")
display(
    high_value_orders[
        ["Order_ID", "City", "Product", "Total_Sales"]
    ].sort_values("Total_Sales", ascending=False)
)

# Orders from Mumbai
print("Mumbai Orders:")
display(
    df[df["City"] == "Mumbai"][
        ["Order_ID", "Product", "Category", "Total_Sales"]
    ]
)

# Sports category orders
print("Sports Category Orders:")
display(
    df[df["Category"] == "Sports"][
        ["Order_ID", "Product", "Quantity", "Total_Sales"]
    ]
)

# UPI payment orders
print("UPI Orders:")
display(
    df[df["Payment_Method"] == "UPI"][
        ["Order_ID", "Product", "Total_Sales"]
    ]
)

# Multiple-condition filtering
print("Sports Orders with Sales > ₹5,000:")
display(
    df[
        (df["Category"] == "Sports") &
        (df["Total_Sales"] > 5000)
    ][
        ["Order_ID", "City", "Product", "Total_Sales"]
    ]
)

# Top 10 orders by sales
print("Top 10 Orders by Sales:")
display(
    df.sort_values(
        by="Total_Sales",
        ascending=False
    ).head(10)
)

# Top 10 orders by quantity
print("Top 10 Orders by Quantity:")
display(
    df.sort_values(
        by="Quantity",
        ascending=False
    ).head(10)
)
```

```python
# ============================================================
# 4. OVERALL SALES ANALYSIS
# ============================================================

total_sales = df["Total_Sales"].sum()
average_sales = df["Total_Sales"].mean()
highest_sales = df["Total_Sales"].max()
lowest_sales = df["Total_Sales"].min()
total_quantity = df["Quantity"].sum()
number_of_orders = df["Order_ID"].nunique()

print(f"Total Sales: ₹{total_sales:,.2f}")
print(f"Average Sales per Order: ₹{average_sales:,.2f}")
print(f"Highest Order Value: ₹{highest_sales:,.2f}")
print(f"Lowest Order Value: ₹{lowest_sales:,.2f}")
print(f"Total Quantity Sold: {total_quantity}")
print(f"Number of Orders: {number_of_orders}")

# Highest-value order
print("\nHighest Value Order:")
display(df.loc[df["Total_Sales"].idxmax()])

# Lowest-value order
print("\nLowest Value Order:")
display(df.loc[df["Total_Sales"].idxmin()])

# Orders above average sales
above_average = df[
    df["Total_Sales"] > df["Total_Sales"].mean()
]

print("\nOrders Above Average Sales:")
display(
    above_average[
        ["Order_ID", "Product", "City", "Total_Sales"]
    ].sort_values(
        by="Total_Sales",
        ascending=False
    ).head(10)
)
```

```python
# ============================================================
# 5. GROUPBY & AGGREGATION ANALYSIS
# ============================================================

# ---------------- CATEGORY-WISE ANALYSIS ----------------

category_analysis = df.groupby("Category").agg(
    Total_Sales=("Total_Sales", "sum"),
    Average_Sales=("Total_Sales", "mean"),
    Highest_Sale=("Total_Sales", "max"),
    Lowest_Sale=("Total_Sales", "min"),
    Total_Quantity=("Quantity", "sum"),
    Number_of_Orders=("Order_ID", "count")
).sort_values(
    by="Total_Sales",
    ascending=False
)

print("Category-wise Analysis:")
display(category_analysis)


# ---------------- CITY-WISE ANALYSIS ----------------

city_analysis = df.groupby("City").agg(
    Total_Sales=("Total_Sales", "sum"),
    Average_Sales=("Total_Sales", "mean"),
    Total_Quantity=("Quantity", "sum"),
    Number_of_Orders=("Order_ID", "count")
).sort_values(
    by="Total_Sales",
    ascending=False
)

print("City-wise Analysis:")
display(city_analysis)


# ---------------- PRODUCT-WISE ANALYSIS ----------------

product_analysis = df.groupby("Product").agg(
    Total_Sales=("Total_Sales", "sum"),
    Average_Sales=("Total_Sales", "mean"),
    Total_Quantity=("Quantity", "sum"),
    Number_of_Orders=("Order_ID", "count")
).sort_values(
    by="Total_Sales",
    ascending=False
)

print("Top 10 Products by Sales:")
display(product_analysis.head(10))


# ---------------- PAYMENT METHOD ANALYSIS ----------------

payment_analysis = df.groupby("Payment_Method").agg(
    Total_Sales=("Total_Sales", "sum"),
    Average_Sales=("Total_Sales", "mean"),
    Total_Quantity=("Quantity", "sum"),
    Number_of_Orders=("Order_ID", "count")
).sort_values(
    by="Total_Sales",
    ascending=False
)

print("Payment Method Analysis:")
display(payment_analysis)


# ---------------- REGION-WISE ANALYSIS ----------------

region_analysis = df.groupby("Region").agg(
    Total_Sales=("Total_Sales", "sum"),
    Average_Sales=("Total_Sales", "mean"),
    Total_Quantity=("Quantity", "sum"),
    Number_of_Orders=("Order_ID", "count")
).sort_values(
    by="Total_Sales",
    ascending=False
)

print("Region-wise Analysis:")
display(region_analysis)
```

```python
# ============================================================
# 6. DATE & TOP-PERFORMER ANALYSIS
# ============================================================

# Convert Order_Date to datetime
df["Order_Date"] = pd.to_datetime(df["Order_Date"])

# Extract month
df["Month"] = df["Order_Date"].dt.month

# Monthly sales
monthly_analysis = df.groupby("Month").agg(
    Total_Sales=("Total_Sales", "sum"),
    Number_of_Orders=("Order_ID", "count"),
    Total_Quantity=("Quantity", "sum")
).sort_values(
    by="Total_Sales",
    ascending=False
)

print("Monthly Sales Analysis:")
display(monthly_analysis)


# Top-performing category
top_category = (
    df.groupby("Category")["Total_Sales"]
    .sum()
    .idxmax()
)

# Top-performing product
top_product = (
    df.groupby("Product")["Total_Sales"]
    .sum()
    .idxmax()
)

# Top-performing city
top_city = (
    df.groupby("City")["Total_Sales"]
    .sum()
    .idxmax()
)

# Top payment method
top_payment = (
    df.groupby("Payment_Method")["Total_Sales"]
    .sum()
    .idxmax()
)

print("Top Performing Category:", top_category)
print("Top Performing Product:", top_product)
print("Top Performing City:", top_city)
print("Top Payment Method:", top_payment)
```
